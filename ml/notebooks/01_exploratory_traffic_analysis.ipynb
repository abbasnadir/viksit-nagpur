# 🚦 Nagpur Traffic AI: Exploratory Data Analysis

This notebook analyzes traffic volume patterns, peak congestion hours, and speed distributions across key Nagpur intersections (Sitabuldi, Law College, Shankar Nagar, RBI Square, Rahate Colony).

```python
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual theme
sns.set_theme(style="darkgrid")
plt.rcParams["figure.figsize"] = (12, 6)
```

## 1. Load Nagpur Junctions GeoJSON

```python
with open("../../data/geojson/nagpur_junctions.geojson", "r") as f:
    junctions_data = json.load(f)

print(f"Loaded {len(junctions_data['features'])} Nagpur junctions.")
```

## 2. Load or Simulate Time-Series Traffic Telemetry

```python
from ml.src.train_predictor import generate_synthetic_historical_data

df = generate_synthetic_historical_data(n_days=7)
df.head()
```

## 3. Hourly Congestion Profile

```python
df["hour"] = df["timestamp"].dt.hour
hourly_summary = df.groupby(["hour", "junction_id"])["volume_pcu"].mean().reset_index()

plt.figure(figsize=(14, 6))
sns.lineplot(data=hourly_summary, x="hour", y="volume_pcu", hue="junction_id", marker="o")
plt.title("Average Hourly Traffic Volume (PCU/hr) across Nagpur Junctions", fontsize=14)
plt.xlabel("Hour of Day (24h)")
plt.ylabel("Passenger Car Units (PCU/hr)")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
